```python
# calculate the average effort for each effort level
def calculate_average_effort(effort_data):
    effort_dict = {}
    lines = effort_data.strip().split("\n")
    for line in lines:
        effort_level, effort_value = line.split(", ")
        effort_dict[effort_level] = float(effort_value)
    return sum(effort_dict.values()) / len(effort_dict)
```
